```r
install.packages("tidyverse")
install.packages("lubridate")
install.packages("janitor")
```

```r
library(tidyverse)
accidents <- read_csv("US_Accidents_March23.csv")
# accidents <- read_csv("US_Accidents_March23_10pct.csv")

# head(accidents)
# glimpse(accidents)
```

```r
library(janitor)

accidents <- accidents %>%
  clean_names()

names(accidents)
```

```r
library(lubridate)

bool_cols <- c("amenity", "bump", "crossing", "give_way", "junction",
               "no_exit", "railway", "roundabout", "station", "stop",
               "traffic_calming", "traffic_signal", "turning_loop")

accidents <- accidents %>%
  mutate(across(all_of(bool_cols), ~ as.logical(.)))
```

```r
accidents <- accidents %>% mutate(accident_duration_min = as.numeric(difftime(end_time, start_time, units = "mins")))
```

```r
essential <- c("severity", "start_lat", "start_lng")

accidents_clean <- accidents %>%
  filter(if_all(all_of(essential), ~ !is.na(.)))

accidents_clean <- accidents_clean %>% distinct(id, .keep_all = TRUE)
```

```r
accidents <- accidents_clean
```
